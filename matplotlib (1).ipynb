```python
import matplotlib.pyplot as plt
```

```python
months=['Jan','feb','mar','apr','may']
sales=[100000,1500000,1200000,200000,2500000]
plt.plot(months,sales,marker="o",mec="red",mfc="yellow",ms=16,linestyle="--",color='green')
plt.show()
```

```python
months=['Jan','feb','mar','apr','may']
sales=[100000,1500000,1200000,200000,2500000]
plt.bar(months,sales,edgecolor='black',color='pink',linewidth=2)
plt.xlabel('Month-wise')
plt.ylabel('sales')
plt.title('months vs sales')
plt.show()
```

```python
months=['Jan','feb','mar','apr','may']
sales=[100000,1500000,1200000,200000,2500000]
plt.barh(months,sales,edgecolor='red')
plt.xlabel('Month-wise')
plt.ylabel('sales')
plt.title('months vs sales')
plt.show()
```
